# Series

A series is a one-dimensional array of indexed data.

A pandas.Series having an index is the main difference between a `pandas.Series` and a NumPy array.

```python
# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

# Creating a 'pandas.Series'

`s = pd.Series(data, index=index)`

The `data` parameter can be:
- a list or NumPy array
- a Python dictionary
- a single number, boolean (True/False), or string

`index` parameter is optional
If included it must be a list of indices of the same length as `data`
If we don't include it, the default is to make the index equal to `[0,...,len(data)-1]`


# Example: Creating a `pandas.Series` from a NumPy array
We need to pass a NumPy array (or a list of objects that can be converted to NumPy types) as `data`

```python
# A series from a numpy array 
pd.Series(np.arange(3), index=[2023, 2024, 2025])
```

Example: Creating a `pandas.Series` from a list

```python
# A series from a list of strings with default index
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

# Example: Creating a pandas.Series from a dictionary
Dictionary: a set of key-value pairs. 

When creating a `pandas.Series` via a dictionary the keys will become the index and the values the corresponding data.

```python
# Construct dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

# Initialize series using a dictionary
pd.Series(d)
```

```python
# An object in pandas usually indicates that the series is made up of strings. 
# Object data type can also be a mix of strings and numbers
```

# Example: Creating a `pandas.Series` from a single value

If we only provide a single number, boolean, or string as the data for the series, we need to provide an index. The value will be repeated to match the length of the index.

```python
pd.Series(3.0, index = ['A', 'B', 'C'])
```

# Simple Operations

Arithmetic operations and most NumPy functions work on series

Example:

```python
# Define a series
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10, '\n')

# Take the exponential of each element in series
print(np.exp(s), '\n') # `\n` prints output to new line

# Original series is unchanged
print(s)
```

```python
# Showing that without `\n' output is printed on top of each other
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10)

# Take the exponential of each element in series
print(np.exp(s))

# Original series is unchanged
print(s)
```

# Identifying Missing Values
We can represent a missing, NULL, or NA value with the float value `numpy.nan`, which stands for "not a number"

```python
# Series with NAs in it
s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# Check if series has NAs
s.hasnans

# The hasnans attribute returns `True` if there are any NA values 
```

```python
# The `isna` method gives as an output the elements in the series that are NA's

s.isna()
```

# Data Frames

`pandas.DataFrame` represents tabular data (like a spreadsheet)
Each column of a `pandas.DataFrame` is a `pandas.Series`.

# Creating a `pandas.DataFrame

A `pandas.DataFrame` is a dictionary of `pandas.Series`, with each column name being the key and the column values being the key's value.

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# We can change the index by changing the index attribute in the data frame
df.index = ['a','b','c']
df
```

# Check-In 2
We can access the data frame’s column names via the columns attribute. Update the column names to C1 and C2 by updating this attribute.

```python
df.columns= ['C1', 'C2']
df
```

# Summary

`pandas.Series` is one dimensional indexed data

`pd.Series(data, index=index)`

If index is left empty the index starts 0, 1, ...
Index provides labels for values

Arithmetic operations and functions can be used on series.

`numpy.nan` (`np.nan`) can be used to represent missing values

`.hasnans` returns True if there are NA values

`.isna()` returns True for NA values, False is not NA

can use a `panda.Series` to create a series then apply p`anda.Dataframe` to create the dataframe from those series

`df.index = ['']`
